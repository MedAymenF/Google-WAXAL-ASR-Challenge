```python
submission_without_punctuation = submission.copy()
submission_without_punctuation['Target'] = submission_without_punctuation['Target'].str.replace(r'[^\w\s]', '', regex=True)
submission_without_punctuation
```

```python
submission_without_punctuation.to_csv("submission_without_punctuation (8).csv", index=False)
```

```python
submission_with_terminal_punctuation = submission.copy()
def add_terminal_punctuation(text):
    if not text.endswith(('.', '!', '?')):
        return text + '.'
    return text

submission_with_terminal_punctuation['Target'] = submission_with_terminal_punctuation['Target'].apply(add_terminal_punctuation)
submission_with_terminal_punctuation
```

```python
submission_with_terminal_punctuation.to_csv("submission_with_terminal_punctuation (8).csv", index=False)
```

```python
import unicodedata
import re

# text = unicodedata.normalize('NFC', text) # Standardizes composed characters
# text = re.sub(r'\s+', ' ', text).strip()  # Fixes rogue spacing
submission_normalized = submission_with_terminal_punctuation.copy()
submission_normalized['Target'] = submission_normalized['Target'].apply(lambda x: unicodedata.normalize('NFC', x))
submission_normalized['Target'] = submission_normalized['Target'].apply(lambda x: re.sub(r'\s+', ' ', x).strip())
submission_normalized.to_csv("submission_nfc_normalized (8).csv", index=False)
```

```python
submission_nfkc_normalized = submission_with_terminal_punctuation.copy()
submission_nfkc_normalized['Target'] = submission_nfkc_normalized['Target'].apply(lambda x: unicodedata.normalize('NFKC', x))
submission_nfkc_normalized['Target'] = submission_nfkc_normalized['Target'].apply(lambda x: re.sub(r'\s+', ' ', x).strip())
submission_nfkc_normalized.to_csv("submission_nfkc_normalized (8).csv", index=False)
```

```python
submission_with_casing_fixed = submission_normalized.copy()

def fix_casing(text):
    """
    Some words have multiple capital letters in the middle of the word, which is not standard. This function fixes that by converting the entire word to lowercase except for the first letter.
    """
    words = text.split()
    for i, word in enumerate(words):
        if word[0].isupper() and len(word) and any(c.isupper() for c in word[1:]):
            print(f"Fixing casing for word: {word}. ", end="")
            words[i] = word[0] + word[1:].lower()
            print(f"Fixed word: {words[i]}")
    return " ".join(words)

submission_with_casing_fixed['Target'] = submission_with_casing_fixed['Target'].apply(fix_casing)
submission_with_casing_fixed.to_csv("submission_with_casing_fixed (8).csv", index=False)
```
